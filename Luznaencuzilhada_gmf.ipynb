```python
from graphviz import Digraph

fluxo = Digraph(
    name='Fluxograma_Espacado',
    format='png',
    engine='neato',
    graph_attr={
        'rankdir': 'LR',
        'labelloc': 't',
        'label': '\nFluxograma do Estudo - Layout Otimizado\n',
        'bgcolor': '#ffffff',
        'fontsize': '16',
        'fontname': 'Arial-Bold',
        'splines': 'polyline',
        'nodesep': '1.0',  # Espaço entre nós
        'ranksep': '2.0',  # Espaço entre níveis
        'sep': '20'        # Espaço geral
    },
    node_attr={
        'style': 'filled',
        'shape': 'Mrecord',
        'align': 'left',
        'fontname': 'Arial',
        'fontsize': '10',
        'penwidth': '1.2',
        'margin': '0.5'    # Margem interna
    },
    edge_attr={
        'color': '#606060',
        'arrowsize': '0.8',
        'penwidth': '1.5',
        'minlen': '3'      # Distância mínima entre nós
    }
)

# Paleta de cores harmonizada
cores = [
    '#E8F5E9', '#FBE9E7', '#E3F2FD',
    '#F3E5F5', '#FFF8E1', '#E0F7FA',
    '#EFEBE9', '#FFECB3'
]

# Etapas com descrições detalhadas
etapas = [
    ("1. Engajamento Inicial",
     "• Mapeamento de terreiros\n• Contato com lideranças\n• TCLE institucional",
     '#C8E6C9'),

    ("2. Adequação Ética",
     "• Adaptação do TCLE\n• Validação comunitária\n• Integração digital",
     '#FFCDD2'),

    ("3. Capacitação",
     "• Workshops presenciais\n• Materiais didáticos\n• Simulações práticas",
     '#BBDEFB'),

    ("4. Coleta Inteligente",
     "• QR Codes estratégicos\n• Dispositivos compartilhados\n• Armazenamento seguro",
     '#E1BEE7'),

    ("5. Triagem",
     "• Análise automatizada\n• Priorização de casos\n• Notificação sigilosa",
     '#FFF3E0'),

    ("6. Acolhimento",
     "• Parceria Clínica Escola\n• Agendamento prioritário\n• Monitoramento inicial",
     '#B2EBF2'),

    ("7. Pesquisa Avançada",
     "• Coleta biológica\n• Análises genômicas\n• Banco de dados",
     '#D7CCC8'),

    ("8. Consolidação",
     "• Relatórios científicos\n• Devolutiva comunitária\n• Publicações",
     '#FFECB3')
]

# Adiciona nós com espaçamento
for i, (titulo, desc, cor) in enumerate(etapas):
    fluxo.node(
        f'e{i}',
        label=f'<<b>{titulo}</b>\n{desc}>',
        fillcolor=cor,
        color='#455A64',
        fontcolor='#212121'
    )

# Layout em grade com posicionamento manual
posicoes = {
    'e0': '0,0!',
    'e1': '3,0!',
    'e2': '6,0!',
    'e3': '9,0!',
    'e4': '12,0!',
    'e5': '15,0!',
    'e6': '18,0!',
    'e7': '21,0!'
}

for no, pos in posicoes.items():
    fluxo.node(no, pos=pos)

# Conexões com setas espaçadas
for i in range(len(etapas)-1):
    fluxo.edge(f'e{i}', f'e{i+1}')

# Elementos especiais com posicionamento
fluxo.node('start', label='Início', pos='-3,0!',
          shape='circle', width='0.6',
          fillcolor='#81C784', style='filled')

fluxo.node('end', label='Conclusão', pos='24,0!',
         shape='circle', width='0.6',
         fillcolor='#EF5350', style='filled')

fluxo.edge('start', 'e0', lhead='cluster_prep')
fluxo.edge('e7', 'end', ltail='cluster_cien')

# Agrupamentos lógicos
with fluxo.subgraph(name='cluster_prep') as c:
    c.attr(style='dashed', color='#78909C', label='Fase Preparatória')
    c.node('e0')
    c.node('e1')

with fluxo.subgraph(name='cluster_oper') as c:
    c.attr(color='#5C6BC0', label='Fase Operacional', bgcolor='#E8EAF6')
    c.node('e2')
    c.node('e3')
    c.node('e4')

with fluxo.subgraph(name='cluster_cien') as c:
    c.attr(color='#00695C', label='Fase Científica', bgcolor='#E0F2F1')
    c.node('e5')
    c.node('e6')
    c.node('e7')

fluxo.attr(overlap='false', outputorder='edgesfirst')
fluxo.render('fluxograma_espacado', view=True, cleanup=True)
```
